## 環境構築

### 階層イメージ図

**① コードブロック（等幅フォントで整列）**

```
my_ecommerce_pipeline (Catalog)
├── bronze   ── 生データ・増分取り込み
├── silver   ── クレンジング済データ
└── gold     ── 分析用・集計データ
```

**② テーブル表現**

| レイヤー | スキーマ | 役割 |
|:--------|:--------|:-----|
| Catalog | `my_ecommerce_pipeline` | プロジェクト全体のカタログ |
| Bronze  | `bronze` | 生データ・増分取り込み |
| Silver  | `silver` | クレンジング済データ |
| Gold    | `gold`   | 分析用・集計データ |

**③ ネストリスト表現**

- **my_ecommerce_pipeline** (Catalog)
  - `bronze` ─ 生データ・増分取り込み
  - `silver` ─ クレンジング済データ
  - `gold` ─ 分析用・集計データ

````sql
-- カタログ、スキーマ、テーブルを作成
-- プロジェクト専用のオリジナルカタログの作成
CREATE CATALOG IF NOT EXISTS my_ecommerce_pipeline;

-- 作成したカタログをデフォルト操作対象に設定
USE CATALOG my_ecommerce_pipeline;

-- カタログ内にレイヤー別のスキーマを作成
CREATE SCHEMA IF NOT EXISTS bronze;
CREATE SCHEMA IF NOT EXISTS silver;
CREATE SCHEMA IF NOT EXISTS gold;

-- 各レイヤー配下にチェックポイント/メタデータ用Volumeを作成
CREATE VOLUME IF NOT EXISTS my_ecommerce_pipeline.bronze.checkpoints;
CREATE VOLUME IF NOT EXISTS my_ecommerce_pipeline.silver.checkpoints;
CREATE VOLUME IF NOT EXISTS my_ecommerce_pipeline.gold.checkpoints;
````
